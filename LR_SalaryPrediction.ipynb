```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
csv_path = "/content/drive/MyDrive/ML/salary_data.csv"
df =pd.read_csv(csv_path)

print(df.head(10))
```

```python
feature_cols = [
    "YearsExperience"
]
target_col= "Salary" #predicted value

x = df[feature_cols].to_numpy(dtype=float)
y = df[target_col].to_numpy(dtype=float)

print("X shape:", x.shape)
print("Y shape:", y.shape)
```

```python
def predict(x, w, b):
    return w * x + b
```

```python
def mse(y_true, y_pred):
    n= len(y_true)
    return (1/n) * np.sum((y_true - y_pred)**2)
```

```python
def gradients(x, y_true, y_pred):
    n = len(x)
    dw = (-2/n) * np.sum(x * (y_true - y_pred)) #
    db = (-2/n) * np.sum(y_true - y_pred) #
    return dw, db
```

```python
w = 0.0
b = 0.0
learningRate = 0.00003 # I chose this rate because in graph it creats more suitable regression line
iterations = 1000


for i in range(iterations):
    y_pred = predict(x, w, b)
    dw, db = gradients(x, y, y_pred)

    # weigts updating
    w = w - learningRate * dw
    b = b - learningRate * db
```

```python
plt.scatter(x, y, label="Datas")
plt.plot(x, predict(x, w, b), color="black", label="LR line")

plt.xlabel("Years Experience")
plt.ylabel("Salary")
plt.title("Linear Regression: Salary Prediction")
plt.legend()
plt.show()
```
